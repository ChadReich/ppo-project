```python
# Imports

import random
import gymnasium as gym
import numpy as np
import torch

# Seed

seed = 42

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)

# Check environments

for env_id in ["CartPole-v1", "Pendulum-v1"]:

    env = gym.make(env_id)

    observation, info = env.reset(seed=seed)
    env.action_space.seed(seed)

    action = env.action_space.sample()

    observation, reward, terminated, truncated, info = env.step(action)

    print(env_id)
    print("Observation shape:", observation.shape)
    print("Action space:", env.action_space)
    print("Reward:", reward)
    print("Terminated:", terminated)
    print("Truncated:", truncated)
    print()

    env.close()
```

using both flags is correct because we are deciding whether the current environment episode needs to be reset. Later, when calculating a bootstrap RL target, we must not do this.
